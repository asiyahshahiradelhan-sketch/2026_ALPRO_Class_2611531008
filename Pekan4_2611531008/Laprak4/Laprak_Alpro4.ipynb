## 2.1 Dasar Teori

**1. Python**
- Python adalah bahasa pemrograman tingkat tinggi yang bersifat *interpreted*, dinamis, dan mudah dipahami.
- Diciptakan oleh Guido van Rossum, pertama kali dirilis tahun 1991.
- Banyak digunakan dalam pengembangan software, data science, AI, dan web.

**2. Struktur Kontrol**

Terdapat 3 jenis struktur kontrol dalam pemrograman:
1. **Sekuensial** — instruksi dijalankan berurutan dari atas ke bawah.
2. **Percabangan (Conditional)** — instruksi dijalankan berdasarkan kondisi tertentu.
3. **Perulangan (Looping)** — instruksi dijalankan berulang selama kondisi terpenuhi.

**3. Conditional Statement**

Struktur yang digunakan untuk mengevaluasi kondisi (`True`/`False`) dan menentukan blok kode yang dijalankan.

- **`if`** → menjalankan kode jika kondisi bernilai `True`.
  ```python
  if kondisi:
      pernyataan
  ```

- **`if-else`** → dua kemungkinan alur (benar/salah).
  ```python
  if kondisi:
      pernyataan_1
  else:
      pernyataan_2
  ```

- **`if-elif-else`** → lebih dari dua kemungkinan kondisi, dicek berurutan.
  ```python
  if kondisi_1:
      pernyataan_1
  elif kondisi_2:
      pernyataan_2
  else:
      pernyataan_3
  ```

- **Nested conditional** → percabangan di dalam percabangan lain.
  ```python
  if kondisi_1:
      if kondisi_2:
          pernyataan
  ```

**4. Operator dalam Conditional**

- **Operator Perbandingan:** `==`, `!=`, `>`, `<`, `>=`, `<=`
- **Operator Logika:**
  - `and` → benar jika kedua kondisi benar
  - `or` → benar jika salah satu kondisi benar
  - `not` → membalik nilai kebenaran

**5. Indentasi**
- Python menggunakan indentasi (bukan `{}`) untuk menandai blok kode.
- Kesalahan indentasi menyebabkan *error*.

### 2.2.3 Program multi_if1 "B_2611531008"

````python
# Buat file dengan nama multi_if1_nim.py
# Buat program untuk kondisional if
# Nama variabel ditambah 4 digit terakhir contok: ipk_1008
# Programi ini menggunakan fungsi input()

umur_1008 = int(input("Input umur anda: "))
sim_1008 = input("Apakah Anda Sudah Punya Sim C (y/t): ")

if umur_1008 >= 17 and sim_1008 == 'y':
    print("Anda Sudah Dewasa dan boleh bawa motor")

if umur_1008 >= 17 and sim_1008 != 'y':
    print("Anda Sudah dewasa tetapi tidak boleh bawa motor")

if umur_1008 < 17 and sim_1008 == 'y':
    print("Anda Belum Cukup umur punya SIM")

if umur_1008 < 17 and sim_1008 != 'y':
    print("Anda Belum cukup umur bawa motor")
````

6. Kondisi 4: Jika umur_1008 < 17 dan sim_1008 != 'y' → mencetak "Anda Belum cukup umur bawa motor" (belum dewasa dan belum punya SIM).

### 2.2.4 Program multi_if2 "B_2611531008"

````python
# Buat file dengan nama multi_if2_nim.py
# buat program untuk kondisional if
# Nama Variabel ditambah 4 digit terakhir contoh: ipk_1008
# Programi ini menggunakan fungsi input()
# Program menghitung diskon Belanja

# Input dan user
total_belanja_1008 = float(input("Input Total Belanja (Rp): "))

# Input status member (mengecek apakah user mengetik 'y atau 't')
input_member_1008 = input("Apakah Anda Member (y/t): ").strip().lower()
is_member_1008 = input_member_1008 in ["y", "t"]

# Input status kode promo (mengecek apakah user mengetik 'y atau 't')
input_promo_1008 = input("Apakah kode promo valid? (y/t): ").strip().lower()
kode_promo_valid_1008 = input_promo_1008 in ["y", "t"]

total_diskon_persen_1008 = 0

# Multi-if terpisah: Setiap kondisi diperiksa secara independen 
# Diskon bisa ditumpuk (akumulasi) jika memenuhi beberapa syarat sekaligus

if total_belanja_1008 > 1000000:
    total_diskon_persen_1008 += 10 # Diskon total besar

if is_member_1008:
    total_diskon_persen_1008 += 5 # Diskon member

if kode_promo_valid_1008:
    total_diskon_persen_1008 += 15 # Diskon voucher

# Menghitung nominal diskon dan total bayar
nominal_diskon_1008 = total_belanja_1008 * (total_diskon_persen_1008 / 100)
total_bayar_1008 = total_belanja_1008 - nominal_diskon_1008

# Output hasil
print("\n--- Rincian pembayaran ---")
print(f"Total diskon  : {total_diskon_persen_1008} (Rp {nominal_diskon_1008:.0f})")
print(f"Total bayar : Rp {total_bayar_1008:.0f}")

print(f"Total diskon yang Anda dapatkan: {total_diskon_persen_1008}%")
# output : Total diskon yang anda dapatkan: 30% jika belanja > 1 juta, member, dan kode promo valid
````

**Penjelasan Program**

1. Input data: Program meminta 3 input — total_belanja_1008 (angka belanja), input_member_1008 (status member), dan input_promo_1008 (status kode promo). Kedua input teks diproses dengan .strip().lower() agar tidak sensitif terhadap huruf besar/kecil atau spasi berlebih.
2. Validasi input menjadi boolean:
is_member_1008 bernilai True jika input_member_1008 berisi "y" atau "t".
kode_promo_valid_1008 bernilai True jika input_promo_1008 berisi "y" atau "t".
3. Struktur multi-if independen (tanpa elif): Ada 3 blok if terpisah yang masing-masing dicek sendiri-sendiri, sehingga diskon bisa ditumpuk/diakumulasi:
Jika total_belanja_1008 > 1.000.000 → tambah diskon 10%.
Jika is_member_1008 bernilai True → tambah diskon 5%.
Jika kode_promo_valid_1008 bernilai True → tambah diskon 15%.
4. Perhitungan akhir: nominal_diskon_1008 dihitung dari persentase total diskon dikali total belanja, lalu total_bayar_1008 = total belanja dikurangi nominal diskon.
5. Output: Menampilkan rincian diskon (persen & nominal rupiah) dan total yang harus dibayar, diformat dengan .0f (tanpa desimal) menggunakan f-string.

5. Kondisi 3 (elif): umur_1008 < 17 dan sim_1008 == 'y' → "Anda Belum Cukup umur punya SIM".
6. Kondisi 4 (else): Menangkap sisa kemungkinan (umur_1008 < 17 dan sim_1008 != 'y') → "Anda Belum cukup umur dan tidak boleh bawa motor".
7. Baris terakhir: print("Program Selesai") selalu dijalankan, terlepas dari kondisi mana yang terpenuhi, karena berada di luar blok if-elif-else.

### 2.2.6 Program nilai "B_2611531008"

````python
# Buat file dengan nama nilai_nim.py
# Buat Program untuk nested if
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1008
# Program ini menggunakan fungsi input()
# program konversi nilai angka menjadi huruf

nilai_1008 = int(input("Input nilai angka: "))

if nilai_1008 >= 81:
    print("A")
elif nilai_1008 >= 70:
    print("B")
elif nilai_1008 >= 60:
    print("C")
elif nilai_1008 >= 50:
    print("D")
else:
    print("E")
````

Jika nilai_1008 >= 81 → cetak "A"
Jika nilai_1008 >= 70 (dan bukan ≥81) → cetak "B"
Jika nilai_1008 >= 60 (dan bukan ≥70) → cetak "C"
Jika nilai_1008 >= 50 (dan bukan ≥60) → cetak "D"
Selain semua itu (nilai_1008 < 50) → cetak "E"

### 2.2.7 Program match_case1 "B_26115310008" 

````python
# Buat file dengan nama match_case1_nim.py
# Buat program untuk match case
# Nama variabel ditamab 4 digit nim terakhir contoh: ipk_1008
# Program ini menggunkaan fungsi input()
# Program konversi angka menjadi nama bulan 

bulan_1008 = int(input("Masukkan angka bulan (1 - 12): "))

match bulan_1008:
    case 1:
        print("Januari")
    case 2:
        print("Februari")
    case  3:
        print("Maret")
    case 4:
        print("April")
    case 5:
        print("Mei")
    case 6:
        print("Juni")
    case 7:
        print("Juli")
    case 8:
        print("Agustus")
    case 9:
        print("September")
    case 10:
        print("oktober")
    case 11:
        print("November")
    case 12:
        print("Desember")
    case _:
        print("Angka tidak valid")
````

**Penjelasan Program**

1. Input data: Program meminta input angka bulan (1–12) dari user melalui input(), dikonversi ke int, dan disimpan di variabel bulan_1008.
2. Struktur match-case: Ini adalah fitur pattern matching yang mirip switch-case di bahasa lain (tersedia sejak Python 3.10). Program mencocokkan nilai bulan_1008 dengan setiap case secara berurutan.
3. Case 1–12: Setiap angka bulan dicocokkan dengan nama bulannya dalam Bahasa Indonesia — 1 → "Januari", 2 → "Februari", dan seterusnya hingga 12 → "Desember".
4. Case _ (wildcard/default): Jika bulan_1008 tidak cocok dengan angka manapun (misal 0, 13, atau angka negatif), maka akan mencetak "Angka tidak valid". Simbol _ di sini berfungsi sebagai kondisi "tangkap semua" (default), setara dengan else pada if-elif-else.

````python
# tugas4_1008.py
# Sistem Loket Terpadu & Audit Transaksi Ekspedisi Wahana

print("=" * 50)
print("SISTEM LOKET TERPADU WAHANA PETUALANGAN")
print("=" * 50)

# Input data pengunjung
nama_1008 = input("Nama Pengunjung : ").strip()
umur_1008 = int(input("Umur : "))
sim_1008 = input("Memiliki SIM C/Kartu Identitas? (y/t) : ").strip().lower()
jumlah_tiket_1008 = int(input("Jumlah Tiket : "))

# If tunggal
if jumlah_tiket_1008 <= 0:
    print("Peringatan: jumlah tiket tidak valid!")

print("\nPilihan Wahana")
print("1. Safari Rimba        (Rp 50.000)")
print("2. Arung Jeram         (Rp 75.000)")
print("3. Motor ATV Ekstrim   (Rp 120.000)")
print("4. Roller Coaster Kilat(Rp 100.000)")
print("5. All-Access VIP      (Rp 220.000)")

paket_1008 = int(input("Pilih Paket (1-5): "))

nama_paket_1008 = ""
harga_satuan_1008 = 0
paket_valid_1008 = True

# Match Case
match paket_1008:
    case 1:
        nama_paket_1008 = "Wahana Safari Rimba"
        harga_satuan_1008 = 50000

    case 2:
        nama_paket_1008 = "Wahana Arung Jeram"
        harga_satuan_1008 = 75000

    case 3:
        nama_paket_1008 = "Wahana Motor ATV Ekstrim"
        harga_satuan_1008 = 120000

    case 4:
        nama_paket_1008 = "Wahana Roller Coaster Kilat"
        harga_satuan_1008 = 100000

    case 5:
        nama_paket_1008 = "Wahana All-Access VIP"
        harga_satuan_1008 = 220000

    case _:
        print("Paket wahana tidak valid!")
        paket_valid_1008 = False

# Proses hanya jika paket valid
if paket_valid_1008:

    # If-Elif-Else dengan operator logika
    if paket_1008 == 3 and umur_1008 >= 17 and sim_1008 == "y":
        print("\nAnda sudah dewasa dan boleh mengendarai ATV sendiri.")

    elif paket_1008 == 3 and umur_1008 >= 17 and sim_1008 != "y":
        print("\nAnda sudah dewasa tetapi tidak boleh membawa motor ATV (wajib didampingi instruktur).")

    elif paket_1008 == 3 and umur_1008 < 17:
        print("\nAnda belum memenuhi syarat usia untuk ATV.")

    total_awal_1008 = harga_satuan_1008 * jumlah_tiket_1008

    # Multi-if terpisah untuk diskon akumulatif
    diskon_1008 = 0

    if total_awal_1008 >= 200000:
        diskon_1008 += total_awal_1008 * 0.10

    if jumlah_tiket_1008 >= 4:
        diskon_1008 += total_awal_1008 * 0.05

    if umur_1008 >= 60:
        diskon_1008 += total_awal_1008 * 0.05

    total_bayar_1008 = total_awal_1008 - diskon_1008

    # Audit Status Operasional
    status_tiket_1008 = jumlah_tiket_1008 > 0
    status_paket_1008 = paket_valid_1008
    status_pembayaran_1008 = total_bayar_1008 > 0

    print("\n" + "=" * 50)
    print("STRUK TRANSAKSI")
    print("=" * 50)
    print("Nama Pengunjung :", nama_1008)
    print("Umur            :", umur_1008)
    print("Paket           :", nama_paket_1008)
    print("Harga Satuan    : Rp", format(harga_satuan_1008, ","))
    print("Jumlah Tiket    :", jumlah_tiket_1008)
    print("Total Awal      : Rp", format(total_awal_1008, ","))
    print("Diskon          : Rp", format(int(diskon_1008), ","))
    print("Total Bayar     : Rp", format(int(total_bayar_1008), ","))

    print("\nAUDIT STATUS OPERASIONAL")
    print("Status Tiket      :", status_tiket_1008)
    print("Status Paket      :", status_paket_1008)
    print("Status Pembayaran :", status_pembayaran_1008)

    print("=" * 50)
    print("Transaksi selesai.")
````

# KESIMPULAN

Berdasarkan praktikum yang telah dilaksanakan mengenai *Conditional pada Python*, dapat disimpulkan bahwa *conditional statement* merupakan struktur kontrol yang digunakan untuk mengevaluasi suatu kondisi dan menentukan blok kode yang akan dieksekusi berdasarkan hasil evaluasi tersebut, yaitu bernilai `True` atau `False`. Python menyediakan beberapa bentuk struktur percabangan, mulai dari `if` yang digunakan untuk mengevaluasi satu kondisi, `if-else` untuk dua kemungkinan alur program, hingga `if-elif-else` yang digunakan untuk mengevaluasi lebih dari dua kondisi secara berurutan. Selain itu, terdapat pula *nested conditional* yang memungkinkan penanganan kondisi yang lebih kompleks dan bertingkat sesuai kebutuhan program.

Dalam penerapannya, penggunaan operator perbandingan seperti `==`, `!=`, `>`, dan `<`, serta operator logika seperti `and`, `or`, dan `not`, sangat berperan penting dalam membentuk ekspresi kondisi yang tepat. Selain itu, ketepatan penulisan indentasi juga menjadi hal yang krusial dalam bahasa pemrograman Python, mengingat kesalahan indentasi dapat menyebabkan program mengalami *error* dan tidak berjalan sebagaimana mestinya.

Secara keseluruhan, melalui praktikum ini pemahaman mengenai logika percabangan tidak hanya diperoleh secara teori, tetapi juga telah diterapkan secara langsung dalam bentuk kode program. Hal ini memberikan manfaat besar dalam melatih kemampuan berpikir logis dan sistematis, yang merupakan bekal penting dalam menyelesaikan berbagai permasalahan pemrograman di masa mendatang.
