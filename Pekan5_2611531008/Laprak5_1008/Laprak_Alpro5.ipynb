## 2.1 Dasar Teori

1. Algoritma
Algoritma adalah urutan langkah-langkah logis dan sistematis yang digunakan untuk menyelesaikan suatu masalah. Dalam pemrograman, algoritma menjadi dasar sebelum suatu program dibuat.
2. Bahasa Pemrograman Python
Python merupakan bahasa pemrograman tingkat tinggi yang memiliki sintaks sederhana sehingga mudah dipelajari. Python banyak digunakan dalam bidang pendidikan, pengembangan aplikasi, analisis data, dan kecerdasan buatan.
3. Perulangan (Looping)
Perulangan adalah struktur kontrol yang digunakan untuk menjalankan suatu perintah secara berulang-ulang. Tujuannya adalah mengurangi penulisan kode yang berulang dan meningkatkan efisiensi program.
4. Perulangan For
Perulangan for digunakan ketika jumlah pengulangan telah diketahui atau dapat ditentukan. Pada Python, perulangan for biasanya menggunakan fungsi range() untuk menentukan banyaknya iterasi. Materi praktikum menunjukkan berbagai contoh penggunaan perulangan for untuk menampilkan angka dan melakukan perhitungan jumlah bilangan.  
5. Nested For
Nested for adalah perulangan yang berada di dalam perulangan lainnya. Teknik ini digunakan untuk membuat pola yang lebih kompleks, seperti pola titik, pola angka, dan pola bintang. Pada praktikum, nested for digunakan untuk menghasilkan berbagai bentuk pola keluaran.

## 2.2 Kode Program

### 2.2.1 Program jumlah_genap "B_2611531008"


```python
# Buat file dengan nama jumlah_genap_NIM.py
# Buat program untuk menghitung jumlah bilangan genap dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1008
# Program ini menggunakan fungsi input()


ulang_1008 = int(input("Masukkan nilai batas: "))

jumlah_genap_1008 = 0
for i_1008 in range(1, ulang_1008 + 1):
    if i_1008 % 2 == 0:
        print(i_1008, end=" ")
        jumlah_genap_1008 = jumlah_genap_1008 + i_1008

        if i_1008 < ulang_1008:
            print(" + ", end="")
        else:
            print(" = ", jumlah_genap_1008, end="")
print()
print("Jumlah =", jumlah_genap_1008)
```

**Penjelasan Program**

1. Program meminta pengguna memasukkan nilai batas.
2. Program melakukan perulangan dari angka 1 hingga nilai batas tersebut.
3. Setiap angka diperiksa apakah merupakan bilangan genap.
4. Jika angka genap, angka tersebut ditampilkan dan ditambahkan ke total penjumlahan.
5. Program juga menampilkan proses penjumlahan menggunakan tanda "+".
6. Setelah semua bilangan genap ditemukan, program menampilkan hasil total penjumlahannya.
7. Program menggunakan **perulangan `for`**, **percabangan `if`**, dan **fungsi `input()`** untuk menjalankan proses tersebut.

**Contoh hasil:**
Jika batas = 10, maka bilangan genap yang dijumlahkan adalah **2 + 4 + 6 + 8 + 10**, sehingga diperoleh **Jumlah = 30**.

### 2.2.2 Program nested_for1 "B_2611531008"

```python
# Buat file dengan nama nested_for1_NIM.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1008
# Program ini menggunakan fungsi input()

batas_1008 = int(input("Masukkan nilai batas: "))
for line_1008 in range(1, batas_1008 + 1):
    for j_1008 in range(1, (-1 * line_1008 + batas_1008) + 1):
        print(".", end="")
    print(line_1008)
```

**Penjelasan Program**

1. Program meminta pengguna memasukkan nilai batas.
2. Nilai batas disimpan dalam variabel batas_1008.
3. Perulangan pertama (line_1008) digunakan untuk menentukan jumlah baris yang akan ditampilkan.
4. Perulangan kedua (j_1008) digunakan untuk mencetak karakter titik (.) pada setiap baris.
5. Jumlah titik yang dicetak akan berkurang satu setiap kali berpindah ke baris berikutnya.
6. Setelah titik selesai dicetak, program menampilkan nomor baris pada akhir baris tersebut.
7. Program menggunakan nested for (perulangan bersarang), yaitu perulangan di dalam perulangan.

### 2.2.3 Program nested_for2 "B_2611531008"

```python
# Buat file dengan nama perulangan_for2_NIM.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terkahir contoh: ulang_1008
# Program ini menggunakan fungsi input()

batas_1008 = int(input("Masukkan nilai batas: "))
for i_1008 in range(1, batas_1008 + 1):
    for j_1008 in range(1, batas_1008 + 1):
        print("*", end=" ")
    print() # pindah ke baris berikutnya
```

### 2.2.5 Program if_elif_else1 "B_2611531008"

```python
# Buat file dengan nama nested_for4_NIM.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1008
# Program ini menggunakan fungsi input()

tinggi_1008 = int(input("Masukkan tinggi pola (bilangan genap, misal 10): "))

if tinggi_1008 % 2 != 0:
    print("Tinggi harus bilangan genap!")
else:
    a_1008 = tinggi_1008
    c_1008 = a_1008
    lebar_1008 = (2 * tinggi_1008 + 1)

    for i_1008 in range(1, tinggi_1008 + 1):
        b_1008 = c_1008 + 1

        for j_1008 in range(1, lebar_1008 + 1):

            
            # Baris atas dan bawah
            if i_1008 == 1 or i_1008 == tinggi_1008:
                if j_1008 == 1 or j_1008 == lebar_1008:
                    print("#", end="")
                else:
                    print("=", end="")

            # Baris isi
            else:
                if j_1008 == 1 or j_1008 == lebar_1008:
                    print("|", end="")
                else:
                    if j_1008 == c_1008:
                        print("<", end="")
                    elif j_1008 == b_1008:
                        print(">", end="")
                    elif j_1008 == (lebar_1008 - c_1008):
                        print("<", end="")
                    elif j_1008 == (lebar_1008  - c_1008 + 1):
                        print(">", end="")
                    elif j_1008 > b_1008 and j_1008 < (lebar_1008 - c_1008):
                        print(".", end="")
                    else:
                        print(" ", end="")

        print()

        # Logika asli Java
        a_1008 -= 2

        if a_1008 <= 0:
            c_1008 = (-a_1008) + 2
        else:
            c_1008 = a_1008
```

**Penjelasan Program**

1. Program meminta pengguna memasukkan tinggi pola yang harus berupa bilangan genap.
2. Program terlebih dahulu memeriksa apakah nilai yang dimasukkan genap atau ganjil.
3. Jika ganjil, program menampilkan pesan kesalahan dan tidak melanjutkan proses.
4. Jika genap, program menghitung lebar pola dan mulai mencetak pola menggunakan nested for (perulangan bersarang).
5. Perulangan luar digunakan untuk mengatur jumlah baris, sedangkan perulangan dalam digunakan untuk mengatur karakter pada setiap kolom.

### 2.2.8 Program perulangan_for3 "B_2611531008"

```python
# Buat file dengan nama perulangan_for3_NIM.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1008
# Program ini menggunakan fungsi input()

ulang_1008 = int(input("Masukkan jumlah perulangan: "))

jumlah_1008 = 0
for i_1008 in range(1, ulang_1008 + 1):
    print(i_1008, end= " ")
    jumlah_1008 = jumlah_1008 + i_1008

    if i_1008 < ulang_1008:
        print("+", end=" ")
    else:
        print("=", jumlah_1008, end=" ")
print()
print("jumlah =", jumlah_1008)
```

### 2.2.9 Program latihan5 "B_2611531008"

```python
# Program perulangan for untuk membuat segitiga

tinggi_1008 = int(input("Masukkan tinggi segitiga: "))
for i_1008 in range(1, tinggi_1008 + 1):
    for j_1008 in range(1, tinggi_1008 - i_1008 + 1):
        print(" ", end="")
    for k_1008 in range(1, 2 * i_1008):
        print("*", end="")
    print()
```

2. Perulangan kedua (j_1008) mencetak spasi di sebelah kiri agar segitiga terlihat berada di tengah.
3. Perulangan ketiga (k_1008) mencetak karakter * pada setiap baris.
4. Jumlah bintang bertambah dua setiap kali berpindah ke baris berikutnya, sehingga membentuk segitiga sama kaki.
5. Setelah satu baris selesai dicetak, program berpindah ke baris berikutnya.

```python
# File : tugas5_1008.py

print("=== PROGRAM JAM PASIR KRISTAL PALINDROMIK (PEKAN 5) ===")

N_1008 = int(input("Masukkan ukuran skala jam pasir (N): "))

# Menghitung lebar border
lebar_1008 = (4 * N_1008) + 5

# Border atas
print("#", end="")
for i_1008 in range(lebar_1008):
    print("=", end="")
print("#")

# Fase 1 : Jam Pasir Atas (N turun sampai 1)
for baris_1008 in range(N_1008, 0, -1):

    print("|", end=" ")

    # Spasi kiri
    for spasi_1008 in range(2 * (N_1008 - baris_1008)):
        print(" ", end="")

    # Angka menurun
    for angka_1008 in range(baris_1008, 0, -1):
        print(angka_1008, end=" ")

    # Kristal tengah
    print("<*>", end=" ")

    # Angka menaik
    for angka_1008 in range(1, baris_1008 + 1):
        print(angka_1008, end=" ")

    # Spasi kanan
    for spasi_1008 in range(2 * (N_1008 - baris_1008)):
        print(" ", end="")

    print("|")

# Fase 2 : Titik Pusat Jam Pasir
print("|", end=" ")

for spasi_1008 in range(2 * N_1008):
    print(" ", end="")

print("<*>", end="")

for spasi_1008 in range(2 * N_1008):
    print(" ", end="")

print(" |")

# Fase 3 : Jam Pasir Bawah (1 naik sampai N)
for baris_1008 in range(1, N_1008 + 1):

    print("|", end=" ")

    # Spasi kiri
    for spasi_1008 in range(2 * (N_1008 - baris_1008)):
        print(" ", end="")

    # Angka menurun
    for angka_1008 in range(baris_1008, 0, -1):
        print(angka_1008, end=" ")

    # Kristal tengah
    print("<*>", end=" ")

    # Angka menaik
    for angka_1008 in range(1, baris_1008 + 1):
        print(angka_1008, end=" ")

    # Spasi kanan
    for spasi_1008 in range(2 * (N_1008 - baris_1008)):
        print(" ", end="")

    print("|")

# Border bawah
print("#", end="")
for i_1008 in range(lebar_1008):
    print("=", end="")
print("#")
```

**Penjelasan Program**
