## 2.1 Dasar Teori
# Landasan Teori

Tipe data (data type) merupakan suatu klasifikasi yang menentukan jenis nilai yang dapat disimpan dalam suatu variabel, beserta operasi apa saja yang dapat dikenakan terhadap nilai tersebut. Tipe data juga menentukan besarnya ruang memori yang dialokasikan untuk menyimpan suatu nilai, sehingga pemilihan tipe data yang tepat berpengaruh langsung terhadap efisiensi penggunaan memori dalam suatu program. Berdampingan dengan konsep tersebut, variabel dapat didefinisikan sebagai suatu identifier atau nama yang digunakan untuk merepresentasikan lokasi penyimpanan data dalam memori komputer, yang nilainya dapat diubah selama proses eksekusi program berlangsung. Setiap variabel memiliki nama, tipe data, dan nilai yang saling berkaitan satu sama lain, dengan penamaan yang umumnya mengikuti kaidah tertentu, seperti tidak boleh diawali dengan angka, tidak mengandung spasi, serta bersifat case-sensitive pada sebagian besar bahasa pemrograman.

Secara umum, tipe data dalam bahasa pemrograman dapat dikelompokkan menjadi beberapa jenis. Tipe data numerik terdiri atas bilangan bulat (*integer*) yang digunakan untuk menyimpan nilai tanpa pecahan, dan bilangan pecahan (*floating point*) yang digunakan untuk menyimpan nilai desimal. Selanjutnya terdapat tipe data karakter (*char*) yang digunakan untuk menyimpan satu buah karakter tunggal, seperti huruf, angka, maupun simbol, serta tipe data string yang digunakan untuk menyimpan rangkaian karakter atau teks. Selain itu, terdapat pula tipe data boolean yang digunakan untuk menyimpan nilai logika dengan dua kemungkinan, yaitu *true* (benar) atau *false* (salah).

Dalam penggunaannya, variabel perlu melalui proses deklarasi, yaitu pendefinisian nama variabel beserta tipe datanya sebelum variabel tersebut digunakan dalam program, serta proses inisialisasi, yaitu pemberian nilai awal terhadap variabel yang telah dideklarasikan. Pada sebagian bahasa pemrograman, kedua proses tersebut dapat dilakukan secara bersamaan dalam satu baris perintah, sedangkan pada bahasa pemrograman lain, deklarasi dan inisialisasi dilakukan secara terpisah. Terkait dengan penggunaan tipe data, dikenal pula konsep type casting atau konversi tipe data, yaitu proses pengubahan suatu nilai dari satu tipe data ke tipe data lain, baik yang dilakukan secara otomatis oleh sistem (*implicit conversion*) maupun yang dilakukan secara sengaja oleh pemrogram (*explicit conversion*). Pemahaman terhadap konsep ini penting untuk menghindari kesalahan hasil perhitungan maupun kesalahan sistem akibat ketidaksesuaian tipe data saat operasi dilakukan.

Selain aspek tipe data dan proses deklarasi-inisialisasi, pemahaman mengenai ruang lingkup variabel (*scope*) juga menjadi bagian penting dalam landasan teori ini. Ruang lingkup variabel menjelaskan area dalam suatu program di mana variabel tersebut dapat diakses dan digunakan. Secara umum, variabel dapat dibedakan menjadi variabel lokal, yang hanya dapat diakses dalam blok atau fungsi tempat variabel tersebut dideklarasikan, dan variabel global, yang dapat diakses di seluruh bagian program. Keseluruhan konsep tersebut saling berkaitan dan menjadi dasar penting yang harus dipahami sebelum mempelajari materi pemrograman yang lebih kompleks.

---

1. Numerik
Tipe data yang digunakan untuk menyimpan nilai berupa angka, baik bilangan bulat (integer) maupun bilangan desimal (float/double), yang dapat diolah melalui operasi matematis.

2. String
Tipe data yang digunakan untuk menyimpan rangkaian karakter atau teks, seperti nama atau kalimat, yang penulisannya diapit tanda kutip (' ' atau " ").

3. Boolean
Tipe data yang digunakan untuk menyimpan nilai logika, yaitu true (benar) atau false (salah), umumnya digunakan dalam percabangan dan perulangan.

4. Input
Proses memasukkan data dari pengguna, biasanya melalui keyboard, untuk disimpan ke dalam variabel agar dapat diproses oleh program.

5. Konstanta
Identifier yang digunakan untuk menyimpan nilai tetap yang tidak dapat diubah selama program berjalan, seperti nilai phi (π).
