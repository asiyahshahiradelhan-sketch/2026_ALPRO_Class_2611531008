## 2.1 Dasar Teori

Python merupakan bahasa pemrograman tingkat tinggi yang bersifat interpretatif, dinamis, dan multi-paradigma, sehingga mendukung gaya pemrograman prosedural, berorientasi objek, maupun fungsional. Sintaksisnya yang ringkas dan mudah dibaca menjadikan Python banyak digunakan dalam berbagai bidang, mulai dari pengembangan aplikasi, analisis data, otomatisasi, hingga kecerdasan buatan. Salah satu konsep dasar yang harus dikuasai sebelum menulis program yang lebih kompleks adalah pemahaman mengenai operator, yaitu simbol yang digunakan untuk memproses satu atau lebih nilai (operand) sehingga menghasilkan suatu nilai baru.

**1. Operator Aritmatika**

Operator aritmatika digunakan untuk melakukan perhitungan matematis terhadap data bertipe numerik, baik integer maupun float. Jenis-jenis operator aritmatika dalam Python adalah sebagai berikut.

- **Penjumlahan (+)**: menjumlahkan dua nilai atau lebih.
- **Pengurangan (–)**: mengurangkan nilai kedua dari nilai pertama.
- **Perkalian (*)**: mengalikan dua nilai.
- **Pembagian (/)**: membagi dua nilai dan menghasilkan bilangan pecahan (float).
- **Pembagian bulat (//)**: membagi dua nilai dan membulatkan hasilnya ke bawah menjadi bilangan bulat.
- **Modulus (%)**: menghasilkan sisa hasil pembagian antara dua nilai.
- **Perpangkatan (**)**: memangkatkan suatu nilai dengan nilai lainnya.

**2. Operator Penugasan**

Operator penugasan digunakan untuk memberikan atau memperbarui nilai suatu variabel. Operator dasar penugasan adalah tanda sama dengan (=). Python juga menyediakan operator penugasan majemuk yang menggabungkan proses aritmatika dengan penugasan dalam satu langkah, di antaranya:

- **+=** : menambahkan nilai lalu menyimpan hasilnya ke variabel yang sama.
- **-=** : mengurangkan nilai lalu menyimpan hasilnya kembali.
- **\*=** : mengalikan nilai lalu memperbarui variabel.
- **/=**, **//=**, **%=**, **\*\*=** : melakukan operasi pembagian, pembagian bulat, modulus, dan perpangkatan dengan cara yang sama, yaitu langsung memperbarui nilai variabel terkait.

**3. Operator Perbandingan**

Operator perbandingan digunakan untuk membandingkan dua buah nilai dan menghasilkan tipe data boolean, yaitu True atau False. Operator ini banyak digunakan dalam struktur pengambilan keputusan seperti percabangan. Jenis-jenisnya meliputi:

- **==** : memeriksa apakah dua nilai sama.
- **!=** : memeriksa apakah dua nilai tidak sama.
- **\>** : memeriksa apakah nilai pertama lebih besar dari nilai kedua.
- **<** : memeriksa apakah nilai pertama lebih kecil dari nilai kedua.
- **\>=** : memeriksa apakah nilai pertama lebih besar atau sama dengan nilai kedua.
- **<=** : memeriksa apakah nilai pertama lebih kecil atau sama dengan nilai kedua.

**4. Operator Logika**

Operator logika digunakan untuk menggabungkan atau membalikkan nilai dari beberapa ekspresi boolean, sehingga sering dipakai untuk menyusun kondisi yang lebih kompleks. Operator logika dalam Python terdiri atas:

- **and** : menghasilkan nilai True apabila seluruh kondisi yang dihubungkan bernilai benar.
- **or** : menghasilkan nilai True apabila salah satu dari kondisi yang dihubungkan bernilai benar.
- **not** : membalikkan nilai kebenaran suatu kondisi, dari True menjadi False atau sebaliknya.

**5. Operator Bitwise**

Operator bitwise digunakan untuk memanipulasi data secara langsung pada level bit (biner), sehingga sebelum diproses, nilai bertipe integer akan dikonversi terlebih dahulu ke dalam bentuk biner. Jenis-jenis operator bitwise meliputi:

- **& (AND)**: menghasilkan bit 1 hanya jika kedua bit yang dibandingkan bernilai 1.
- **| (OR)**: menghasilkan bit 1 apabila salah satu atau kedua bit bernilai 1.
- **^ (XOR)**: menghasilkan bit 1 apabila kedua bit yang dibandingkan berbeda nilai.
- **~ (NOT)**: membalikkan seluruh nilai bit (komplemen satu).
- **<< (Shift Left)**: menggeser seluruh bit ke arah kiri sejumlah posisi tertentu, yang secara matematis setara dengan mengalikan nilai dengan pangkat dua.
- **\>\> (Shift Right)**: menggeser seluruh bit ke arah kanan sejumlah posisi tertentu, yang setara dengan membagi nilai dengan pangkat dua.

**6. Operator Keanggotaan dan Identitas**

Selain kelima jenis operator di atas, Python juga menyediakan dua kelompok operator lain yang tidak kalah penting dalam pemrosesan data, yaitu:

- **Operator keanggotaan (in dan not in)**: digunakan untuk memeriksa ada atau tidaknya suatu nilai di dalam struktur data seperti list, tuple, atau string.
- **Operator identitas (is dan is not)**: digunakan untuk memeriksa apakah dua variabel menunjuk pada objek yang sama persis di dalam memori, bukan sekadar memiliki nilai yang sama.


1. Numerik
Tipe data yang digunakan untuk menyimpan nilai berupa angka, baik bilangan bulat (integer) maupun bilangan desimal (float/double), yang dapat diolah melalui operasi matematis.

2. String
Tipe data yang digunakan untuk menyimpan rangkaian karakter atau teks, seperti nama atau kalimat, yang penulisannya diapit tanda kutip (' ' atau " ").

3. Boolean
Tipe data yang digunakan untuk menyimpan nilai logika, yaitu true (benar) atau false (salah), umumnya digunakan dalam percabangan dan perulangan.

4. Input
Proses memasukkan data dari pengguna, biasanya melalui keyboard, untuk disimpan ke dalam variabel agar dapat diproses oleh program.

5. Konstanta
Identifier yang digunakan untuk menyimpan nilai tetap yang tidak dapat diubah selama program berjalan, seperti nilai phi (π).

### 2.2.1 Program Arimatika_2611531008

```python
# Buat file dengan nama aritmatika_NIM.py
# Buat program untuk operator aritmatika dalam Python
# Nama variabel ditambah 4 digit nim terakhhir contoh: angka1_1008
# Program ini menggunakan fungsi input()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer

angka1_1008 = int(input("Input angka-1: "))
angka2_1008 = int(input("Input angka-2:" ))

# Penjumlahan
hasil = angka1_1008 + angka2_1008
print("\nOperator Penjumlahan")
print("Hasil =", hasil)

# Pengurangan
hasil = angka1_1008 - angka2_1008
print("\nOperator Pengurangan")
print("Hasil =", hasil)

# Perkalian
hasil = angka1_1008 * angka2_1008
print("\nOperator Perkalian")
print("Hasil =", hasil)

# Pembagian, pembagian bulat, dan sisa bagi
if angka2_1008 != 0:
    hasil = angka1_1008 / angka2_1008
    print("\nOperator Pembagian")
    print("Hasil =", hasil)

    hasil = angka1_1008 // angka2_1008
    print("\nOperator Pembagian Bulat")
    print("Hasil =", hasil)

    hasil = angka1_1008 % angka2_1008
    print("\nOperator Sisa Bagi")
    print("Hasil =", hasil)
else:
    print("Angka kedua tidak boleh bernilai 0.")

# Pangkat
hasil = angka1_1008 ** angka2_1008
print("\nOperator Pangkat")
print("Hasil =", hasil)
```

**Penjelasan Program 2.2.1:**

1. Program diawali dengan memasukkan dua buah angka menggunakan fungsi `input()`.
2. Karena hasil dari `input()` selalu berbentuk teks (string), saya menambahkan fungsi `int()` di depannya agar angka yang dimasukkan bisa langsung diproses sebagai bilangan bulat.
3. Kedua nilai tersebut disimpan pada variabel `angka1_1008` dan `angka2_1008`, di mana angka 1008 di belakang nama variabel merupakan empat digit terakhir NIM saya sesuai ketentuan tugas.
4. Untuk operasi penjumlahan (+), pengurangan (–), dan perkalian (\*), saya langsung menjalankannya tanpa syarat tambahan karena ketiga operasi ini tidak berisiko menimbulkan error.
5. Pada bagian pembagian, saya menambahkan pengecekan `if angka2_1008 != 0:` sebelum menjalankan operator pembagian (/), pembagian bulat (//), dan sisa bagi (%).
6. Pengecekan tersebut saya buat karena saat mencoba-coba sebelumnya, program sempat mengalami error saat angka kedua diisi 0. Setelah dicari tahu, ternyata pembagian dengan angka 0 di Python akan memunculkan error bernama `ZeroDivisionError`.
7. Jika angka kedua bernilai 0, program tidak akan menjalankan proses pembagian, melainkan hanya menampilkan pesan "Angka kedua tidak boleh bernilai 0."
8. Untuk operasi pangkat (\*\*), saya letakkan di luar kondisi if karena operasi ini tidak berkaitan dengan pembagian, sehingga tetap aman dijalankan meskipun angka kedua bernilai 0.

### 2.2.2 Program Assigment_26111531008

```python
# Buat file dengan nama assigment_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka_1008
# Program ini menggunakan fungsi input()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
# Program operator assigment dalam Python

angka1_1008 = int(input("Input angka-1: "))
angka2_1008 = int(input("Input angka-2: "))

print("\nNilai awal angka =", angka1_1008)
print("\nNilai awal angka =", angka2_1008)

# Assigment biasa
hasil = angka1_1008
print("\nAssigment biasa (=)")
print("Hasil =", hasil)

# Assigment penambahan
hasil = angka1_1008
hasil += angka2_1008
print("\nAssigment penambahan (+=)")
print("Hasil =", hasil)

# Assigment pengurangan
hasil = angka1_1008
hasil -= angka2_1008
print("\nAssigment pengurangan (-=)")
print("Hasil =", hasil)

# Assigment perkalian
hasil = angka1_1008
hasil *= angka2_1008
print("\nAssigment perkalian (*=)")
print("Hasil =", hasil)

# Assigment pembagian, pembagian bulat, dan sisa bagi
if angka2_1008 != 0:
    hasil = angka1_1008
    hasil /= angka2_1008
    print("\nAssigment pembagian bulat (/=)")
    print("Hasil =", hasil)
    # Operator tambahan
    hasil = angka1_1008
    hasil //= angka2_1008
    print("\nAssigment pembagian bulat (//=)")
    print("Hasil =", hasil)
    hasil = angka1_1008
    hasil %= angka2_1008
    print("\nAssigment sisa bagi (%=)")
    print("Hasil =", hasil)
else:
    print("\nPembagian tidak dapat dilakukan ")
    print("Angka kedua tidak boleh bernilai 0")

# Operator tambahan: assigment perpangkatan
hasil = angka1_1008
hasil **= angka2_1008
print("\nAssigment perpangkatan (**=)")
print("Hasil =", hasil)
```

**Penjelasan Program 2.2.2**

1. Program diawali dengan memasukkan dua buah angka menggunakan fungsi `input()`, yang kemudian dibungkus dengan `int()` agar nilainya langsung dikonversi menjadi tipe data integer.

3. Setelah input dimasukkan, nilai awal dari kedua variabel tersebut ditampilkan terlebih dahulu menggunakan `print()`, sebagai pembanding sebelum dilakukan proses assignment.
4. Pada bagian **assignment biasa (=)**, variabel `hasil` langsung diisi dengan nilai `angka1_1008` tanpa ada operasi tambahan, hanya untuk menunjukkan cara kerja dasar operator sama dengan.
5. Pada bagian **assignment penambahan (+=)**, variabel `hasil` di-reset dahulu ke `angka1_1008`, kemudian dijumlahkan dengan `angka2_1008` menggunakan `+=`, sehingga sama saja artinya dengan `hasil = hasil + angka2_1008`.
6. Cara yang sama juga diterapkan pada **assignment pengurangan (-=)** dan **assignment perkalian (\*=)**, di mana variabel `hasil` selalu di-reset terlebih dahulu ke `angka1_1008` sebelum dioperasikan, supaya setiap hasil yang muncul murni berasal dari `angka1_1008` dan `angka2_1008`, bukan hasil dari operasi sebelumnya.

Hal ini dilakukan karena ketiga operator tersebut melibatkan proses pembagian, sehingga jika `angka2_1008` bernilai 0, program akan mengalami error bernama `ZeroDivisionError`.
8. Apabila angka kedua bernilai 0, maka program tidak akan menjalankan proses pembagian sama sekali, melainkan hanya menampilkan pesan bahwa pembagian tidak dapat dilakukan karena angka kedua tidak boleh bernilai 0.
9. Terakhir, pada bagian **assignment perpangkatan (\*\*=)**, variabel `hasil` kembali di-reset ke `angka1_1008` lalu dipangkatkan dengan `angka2_1008`. Bagian ini saya letakkan di luar kondisi if karena tidak berkaitan dengan pembagian, sehingga tetap aman dijalankan meskipun angka kedua bernilai 0.

### 2.2.3 Program Bitwise_2611531008

```python
# Buat file dengan nama bitwise_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1008
# Program ini menggunakan fungsi input()

print("======================================")
print("3. OPERATOR BITWISE")
print("======================================")

angka1_1008 = int(input("Masukkan angka bitwise-1: "))
angka2_1008 = int(input("Masukkan angka bitwise-2: "))

print("\nAngka dalam bentuk desimal dan biner")
print("angka1 =",angka1_1008,"| biner =",bin(angka1_1008))
print("angka1 =",angka2_1008,"| biner =",bin(angka2_1008))

# Bitwise AND
hasil_1008 = angka1_1008 & angka2_1008
print("\nBitwise AND (&)")
print(angka1_1008,"&",angka2_1008,hasil_1008)
print("Biner hasil =",bin(hasil_1008))
print("Biner hasil (8 bit) =",format(hasil_1008,"08b"))

# Bitwise OR
hasil_1008 = angka1_1008 | angka2_1008
print("\nBitwise OR (|)")
print(angka1_1008,"|",angka2_1008,hasil_1008)
print("Biner hasil =",bin(hasil_1008))
print("Biner hasil (8 bit) =",format(hasil_1008,"08b"))

# Bitwise XOR
hasil_1008 = angka1_1008 ^ angka2_1008
print("\nBitwise XOR (^)")
print(angka1_1008,"^",angka2_1008,hasil_1008)
print("Biner hasil =",bin(hasil_1008))
print("Biner hasil (8 bit) =",format(hasil_1008,"08b"))

# Bitwise NOT
hasil_1008 = ~angka1_1008
print("\nBitwise NOT (~)")
print(angka1_1008,"~",angka2_1008,hasil_1008)
print("Biner hasil =",bin(hasil_1008))
print("Biner hasil (8 bit) =",format(hasil_1008,"08b"))

# Bitwise geser kiri
jumlah_geser_1008 = int(input("\nMasukkan jumlah pergeseran bit: "))

hasil_1008 = angka1_1008 << jumlah_geser_1008
print("\nBitwise geser kiri (<<)")
print(angka1_1008,"<<",jumlah_geser_1008,"=",hasil_1008)
print("Biner hasil =",bin(hasil_1008))
print("Biner hasil (8 bit) =",format(hasil_1008,"08b"))

# Bitwise geser kanan
hasil_1008 = angka1_1008 >> jumlah_geser_1008
print("\nBitwise geser kanan (>>)")
print(angka1_1008,">>",jumlah_geser_1008,"=",hasil_1008)
print("Biner hasil =",bin(hasil_1008))
print("Biner hasil (8 bit) =",format(hasil_1008,"08b"))
```

**Penjelasan Program 2.2.3**

1. Program diawali dengan memasukkan dua buah angka menggunakan fungsi `input()`, yang kemudian dikonversi menjadi integer dan disimpan pada variabel `angka1_1008` dan `angka2_1008`, dengan angka 1008 menunjukkan empat digit terakhir NIM saya.
2. Sebelum masuk ke operasi bitwise, kedua angka tersebut ditampilkan dalam bentuk desimal sekaligus bentuk binernya menggunakan fungsi `bin()`, supaya saya bisa membandingkan langsung antara angka aslinya dengan representasi bitnya.
3. Pada bagian **Bitwise AND (&)**, kedua angka dibandingkan bit per bit, di mana hasilnya akan bernilai 1 hanya jika kedua bit yang dibandingkan sama-sama bernilai 1. Hasilnya saya tampilkan dalam bentuk desimal, biner biasa (`bin()`), dan biner 8 bit menggunakan `format(hasil,"08b")` agar lebih mudah dibaca posisinya.
4. Pada bagian **Bitwise OR (|)**, cara kerjanya mirip dengan AND, hanya saja hasilnya akan bernilai 1 apabila salah satu atau kedua bit yang dibandingkan bernilai 1.
5. Pada bagian **Bitwise XOR (^)**, hasilnya akan bernilai 1 apabila kedua bit yang dibandingkan berbeda nilai, dan akan bernilai 0 apabila kedua bit tersebut sama.
6. Pada bagian **Bitwise NOT (~)**, operator ini hanya membutuhkan satu operand saja, yaitu `angka1_1008`, karena fungsinya adalah membalikkan seluruh nilai bit dari angka tersebut (dari 0 menjadi 1 dan sebaliknya). Karena itu hasil dari operator NOT sering muncul dalam bentuk negatif, karena Python menyimpan bilangan negatif menggunakan representasi komplemen dua.
7. Setelah bagian NOT, program meminta input tambahan berupa jumlah pergeseran bit yang disimpan pada variabel `jumlah_geser_1008`, yang akan digunakan pada dua operasi selanjutnya.
8. Pada bagian **Bitwise geser kiri (<<)**, seluruh bit dari `angka1_1008` digeser ke kiri sebanyak `jumlah_geser_1008` posisi. Secara matematis, operasi ini sama dengan mengalikan angka tersebut dengan 2 pangkat jumlah pergeserannya.

### 2.2.4 Program Lainnya_2611531008

### 2.2.5 Program Logika_2611531008

```python
# Buat file dengan nama logika_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: a1_1008
# Program ini menggunakan fungsi input()
# Program operator logika dalam Python

# Masukkan nilai boolean
# Input tidak peka terhadap huruf besar dan kecil
a1_1008 = input("Input nilai boolean-1 (true/false): ").strip().lower() == "true"
a2_1008 = input("Input nilai boolean-2 (ture/false): ").strip().lower() == "true"

print("\nA1 =", a1_1008)
print("\nA2 =", a2_1008)

# Konjugsi: bernilai True jika keduanya True
hasil = a1_1008 and a2_1008
print("\nKonjugsi (AND)")
print("A1 and A2 =", hasil)

# Disjungsi: bernilai True jika salah satunya True
hasil = a1_1008 or a2_1008
print("\nDisjungsi (OR)")
print("not A1 or A2 =", hasil)

# Negasi A1: membalik nilai A1
hasil = not a1_1008
print("\nNegasi A1 (NOT)")
print("not A1 =", hasil)

# Negasi A2: membalik nilai A2
hasil = not a2_1008
print("\nNegasi A2 (NOT)")
print("not A2 =", hasil)

# XOR: bernilai True jika kedua nilai berbeda
hasil = a1_1008 != a2_1008
print("\nDisjungsi Eksklusif (XOR)")
print("A1 XOR A2 =", hasil)
```

Karena Python tidak memiliki operator XOR khusus untuk tipe data boolean, cara ini digunakan sebagai alternatif, di mana hasilnya akan bernilai `True` apabila kedua nilai `a1_1008` dan `a2_1008` berbeda, dan bernilai `False` apabila kedua nilai tersebut sama.

###  2.2.6 Program Perbandingan_2611531008

```python
# Buat file dengan nama perbandingan_NIM.py
# Nama variabel ditambah dengan 4 digit nim terakhir contoh: angka1_1008
# Program ini menggunakan fungsi input()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
# Program operator perbandingan dalam Python

angka1_1008 = int(input("Input angka-1: "))
angka2_1008 = int(input("Input angka-2: "))

# Lebih besar dari
hasil = angka1_1008 > angka2_1008
print("\nOperator lebih besar dari")
print("angka1_1008 > angka2_1008 =", hasil)

# Lebih kecil dari
hasil = angka1_1008 < angka2_1008
print("\nOperator lebih kecil dari")
print("angka1_1008 < angka2_1008 =", hasil)

# lebih besar dari atau sama dengan
hasil = angka1_1008 >= angka2_1008
print("\nOperator lebih besar dari atau sama dengan")
print("angka1_1008 >= angka2_1008 =", hasil)

# Lebih kecil dari atau sama dengan
hasil = angka1_1008 <= angka2_1008
print("\nOperator Lebih kecil dari atau sama dengan")
print("angka1_1008 <= angka2_1008 =", hasil)

# Sama dengan
hasil = angka1_1008 == angka2_1008
print("\nOperator Sama dengan")
print("angka1_1008 == angka2_1008 =", hasil)

# Tidak sama dengan
hasil = angka1_1008 != angka2_1008
print("\nOperator Tidak sama dengan")
print("angka1_1008 != angka2_1008 =", hasil)

# Tambahan perbandingan berantai dalam Python
hasil = 0 < angka1_1008 < 100
print("\nPerbandingan berantai")
print("0 < angka1 < 100 =", hasil)

hasil = 0 < angka2_1008 < 100
print("0 < angka2 < 100 =", hasil)
```
