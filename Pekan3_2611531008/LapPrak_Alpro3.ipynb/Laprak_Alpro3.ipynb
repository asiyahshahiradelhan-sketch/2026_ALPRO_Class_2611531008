## 2.1 Dasar Teori
# Landasan Teori

Python adalah bahasa pemrograman tingkat tinggi yang bersifat interpretatif, dinamis, dan mendukung berbagai paradigma pemrograman seperti prosedural, berorientasi objek, dan fungsional. Sintaksisnya yang sederhana membuat Python banyak digunakan dalam pengembangan aplikasi, analisis data, dan kecerdasan buatan.

**1. Operator dalam Python**

Operator adalah simbol khusus yang digunakan untuk melakukan operasi terhadap satu atau lebih nilai (operand) guna menghasilkan nilai baru. Python mengelompokkan operator ke dalam beberapa jenis sebagai berikut.

**2 Operator Aritmatika**

Digunakan untuk melakukan perhitungan matematis antarnilai numerik. Operator ini meliputi penjumlahan (+), pengurangan (–), perkalian (*), pembagian (/), pembagian bulat (//), sisa bagi atau modulus (%), dan perpangkatan (**).

**3. Operator Penugasan**

Digunakan untuk memberikan nilai ke dalam suatu variabel. Operator dasar penugasan adalah tanda sama dengan (=), yang dapat dikombinasikan dengan operator aritmatika sehingga membentuk operator penugasan majemuk seperti +=, -=, *=, dan /=.

**4. Operator Perbandingan**

Digunakan untuk membandingkan dua nilai dan menghasilkan nilai boolean (True atau False). Operator ini meliputi sama dengan (==), tidak sama dengan (!=), lebih besar (>), lebih kecil (<), lebih besar sama dengan (>=), dan lebih kecil sama dengan (<=).

**5. Operator Logika**

Digunakan untuk menggabungkan atau membalikkan hasil dari ekspresi bertipe boolean. Operator logika terdiri atas and (menghasilkan True jika kedua kondisi benar), or (menghasilkan True jika salah satu kondisi benar), dan not (membalikkan nilai kebenaran suatu kondisi).

**6. Operator Keanggotaan (Membership)**

Digunakan untuk memeriksa keberadaan suatu nilai dalam struktur data seperti list, tuple, atau string. Operator ini terdiri atas in dan not in.

**7. Operator Identitas**

Digunakan untuk memeriksa apakah dua variabel merujuk pada objek yang sama di dalam memori, menggunakan operator is dan is not.

### 2.2.1 Program Aritmatika "B_2611531008"


```python
# Buat file dengan nama aritmatika_NIM.py
# Buat program untuk operator aritmatika dalam Python
# Nama variabel ditambah 4 digit nim terakhhir contoh: angka1_1008
# Program ini menggunakan fungsi input()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer

angka1_1008 = int(input("Input angka-1: "))
angka2_1008 = int(input("Input angka-2:" ))

# Penjumlahan
hasil = angka1_1008 + angka2_1008
print("\nOperator Penjumlahan")
print("Hasil =", hasil)

# Pengurangan
hasil = angka1_1008 - angka2_1008
print("\nOperator Pengurangan")
print("Hasil =", hasil)

# Perkalian
hasil = angka1_1008 * angka2_1008
print("\nOperator Perkalian")
print("Hasil =", hasil)

# Pembagian, pembagian bulat, dan sisa bagi
if angka2_1008 != 0:
    hasil = angka1_1008 / angka2_1008
    print("\nOperator Pembagian")
    print("Hasil =", hasil)

    hasil = angka1_1008 // angka2_1008
    print("\nOperator Pembagian Bulat")
    print("Hasil =", hasil)

    hasil = angka1_1008 % angka2_1008
    print("\nOperator Sisa Bagi")
    print("Hasil =", hasil)
else:
    print("Angka kedua tidak boleh bernilai 0.")

# Pangkat
hasil = angka1_1008 ** angka2_1008
print("\nOperator Pangkat")
print("Hasil =", hasil)
```

**Penjelasan Program**

2. Operator penjumlahan (+), pengurangan (–), dan perkalian (\*) dijalankan langsung karena tidak berisiko error.
3. Operator pembagian (/), pembagian bulat (//), dan sisa bagi (%) ditempatkan di dalam `if angka2_1008 != 0:` untuk mencegah error `ZeroDivisionError` saat angka kedua bernilai 0.
4. Jika angka kedua bernilai 0, program menampilkan pesan peringatan tanpa menjalankan proses pembagian.
5. Operator pangkat (\*\*) diletakkan di luar kondisi if karena tidak berkaitan dengan pembagian, sehingga aman dijalankan meskipun angka kedua bernilai 0.
6. Setiap hasil operasi ditampilkan menggunakan `print()` sesuai jenis operatornya.


### 2.2.2 Program Assigment "B_2611531008"

```python
# Buat file dengan nama assigment_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka_1008
# Program ini menggunakan fungsi input()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
# Program operator assigment dalam Python

angka1_1008 = int(input("Input angka-1: "))
angka2_1008 = int(input("Input angka-2: "))

print("\nNilai awal angka =", angka1_1008)
print("\nNilai awal angka =", angka2_1008)

# Assigment biasa
hasil = angka1_1008
print("\nAssigment biasa (=)")
print("Hasil =", hasil)

# Assigment penambahan
hasil = angka1_1008
hasil += angka2_1008
print("\nAssigment penambahan (+=)")
print("Hasil =", hasil)

# Assigment pengurangan
hasil = angka1_1008
hasil -= angka2_1008
print("\nAssigment pengurangan (-=)")
print("Hasil =", hasil)

# Assigment perkalian
hasil = angka1_1008
hasil *= angka2_1008
print("\nAssigment perkalian (*=)")
print("Hasil =", hasil)

# Assigment pembagian, pembagian bulat, dan sisa bagi
if angka2_1008 != 0:
    hasil = angka1_1008
    hasil /= angka2_1008
    print("\nAssigment pembagian bulat (/=)")
    print("Hasil =", hasil)
    # Operator tambahan
    hasil = angka1_1008
    hasil //= angka2_1008
    print("\nAssigment pembagian bulat (//=)")
    print("Hasil =", hasil)
    hasil = angka1_1008
    hasil %= angka2_1008
    print("\nAssigment sisa bagi (%=)")
    print("Hasil =", hasil)
else:
    print("\nPembagian tidak dapat dilakukan ")
    print("Angka kedua tidak boleh bernilai 0")

# Operator tambahan: assigment perpangkatan
hasil = angka1_1008
hasil **= angka2_1008
print("\nAssigment perpangkatan (**=)")
print("Hasil =", hasil)
```

**Penjelasan Program**

1. Program meminta pengguna memasukkan dua angka melalui `input()`, dikonversi menjadi integer, dan disimpan pada variabel `angka1_1008` dan `angka2_1008`, lalu nilai awal keduanya ditampilkan sebagai pembanding.
2. Pada setiap operasi, variabel `hasil` selalu di-reset kembali ke `angka1_1008` sebelum dioperasikan, agar setiap hasil murni berasal dari `angka1_1008` dan `angka2_1008`, bukan hasil dari operasi sebelumnya.
3. Operator assignment biasa (=) hanya menyalin nilai `angka1_1008` ke `hasil` tanpa operasi tambahan.
4. Operator penambahan (+=), pengurangan (-=), dan perkalian (\*=) dijalankan langsung karena tidak berisiko error.
5. Operator pembagian (/=), pembagian bulat (//=), dan sisa bagi (%=) ditempatkan di dalam `if angka2_1008 != 0:` untuk mencegah error `ZeroDivisionError` saat angka kedua bernilai 0.
6. Jika angka kedua bernilai 0, program menampilkan pesan bahwa pembagian tidak dapat dilakukan tanpa menjalankan proses pembagian.
7. Operator perpangkatan (\*\*=) diletakkan di luar kondisi if karena tidak berkaitan dengan pembagian, sehingga aman dijalankan meskipun angka kedua bernilai 0.
8. Setiap operator assignment majemuk (+=, -=, \*=, dst.) pada dasarnya merupakan bentuk singkat dari operasi biasa yang langsung digabung dengan penyimpanan nilai ke variabel yang sama.

### 2.2.3 Program Bitwise "B_2611531008"

```python
# Buat file dengan nama bitwise_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1008
# Program ini menggunakan fungsi input()

print("======================================")
print("3. OPERATOR BITWISE")
print("======================================")

angka1_1008 = int(input("Masukkan angka bitwise-1: "))
angka2_1008 = int(input("Masukkan angka bitwise-2: "))

print("\nAngka dalam bentuk desimal dan biner")
print("angka1 =",angka1_1008,"| biner =",bin(angka1_1008))
print("angka1 =",angka2_1008,"| biner =",bin(angka2_1008))

# Bitwise AND
hasil_1008 = angka1_1008 & angka2_1008
print("\nBitwise AND (&)")
print(angka1_1008,"&",angka2_1008,hasil_1008)
print("Biner hasil =",bin(hasil_1008))
print("Biner hasil (8 bit) =",format(hasil_1008,"08b"))

# Bitwise OR
hasil_1008 = angka1_1008 | angka2_1008
print("\nBitwise OR (|)")
print(angka1_1008,"|",angka2_1008,hasil_1008)
print("Biner hasil =",bin(hasil_1008))
print("Biner hasil (8 bit) =",format(hasil_1008,"08b"))

# Bitwise XOR
hasil_1008 = angka1_1008 ^ angka2_1008
print("\nBitwise XOR (^)")
print(angka1_1008,"^",angka2_1008,hasil_1008)
print("Biner hasil =",bin(hasil_1008))
print("Biner hasil (8 bit) =",format(hasil_1008,"08b"))

# Bitwise NOT
hasil_1008 = ~angka1_1008
print("\nBitwise NOT (~)")
print(angka1_1008,"~",angka2_1008,hasil_1008)
print("Biner hasil =",bin(hasil_1008))
print("Biner hasil (8 bit) =",format(hasil_1008,"08b"))

# Bitwise geser kiri
jumlah_geser_1008 = int(input("\nMasukkan jumlah pergeseran bit: "))

hasil_1008 = angka1_1008 << jumlah_geser_1008
print("\nBitwise geser kiri (<<)")
print(angka1_1008,"<<",jumlah_geser_1008,"=",hasil_1008)
print("Biner hasil =",bin(hasil_1008))
print("Biner hasil (8 bit) =",format(hasil_1008,"08b"))

# Bitwise geser kanan
hasil_1008 = angka1_1008 >> jumlah_geser_1008
print("\nBitwise geser kanan (>>)")
print(angka1_1008,">>",jumlah_geser_1008,"=",hasil_1008)
print("Biner hasil =",bin(hasil_1008))
print("Biner hasil (8 bit) =",format(hasil_1008,"08b"))
```

**Penjelasan Program**

1. Program meminta pengguna memasukkan dua angka melalui `input()`, dikonversi menjadi integer, dan disimpan pada variabel `angka1_1008` dan `angka2_1008`.
2. Kedua angka ditampilkan dalam bentuk desimal dan biner menggunakan fungsi `bin()` sebagai pembanding sebelum dilakukan operasi bitwise.
3. Operator AND (&) membandingkan bit per bit kedua angka, hasilnya 1 hanya jika kedua bit bernilai 1.
4. Operator OR (|) menghasilkan bit 1 apabila salah satu atau kedua bit bernilai 1.
5. Operator XOR (^) menghasilkan bit 1 apabila kedua bit yang dibandingkan berbeda nilai.
6. Operator NOT (~) hanya menggunakan satu operand (`angka1_1008`), berfungsi membalik seluruh nilai bit sehingga hasilnya sering berupa bilangan negatif akibat representasi komplemen dua di Python.
7. Setiap hasil operasi ditampilkan dalam tiga bentuk: desimal, biner (`bin()`), dan biner 8 bit (`format(hasil,"08b")`) agar perubahan bit lebih mudah dibaca.
8. Program meminta input tambahan berupa jumlah pergeseran bit (`jumlah_geser_1008`) yang digunakan pada operator geser kiri dan kanan.
9. Operator geser kiri (<<) menggeser seluruh bit ke kiri, setara dengan mengalikan angka dengan 2 pangkat jumlah geser.
10. Operator geser kanan (>>) menggeser seluruh bit ke kanan, setara dengan membagi angka dengan 2 pangkat jumlah geser.


### 2.2.4 Program Lainnya "B_2611531008"

```python
# Buat file dengan nama lainnya_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1008
# Program ini menggunakan fungsi input()
# Program operator keanggotaan dan identitas

print("======================================")
print("1. OPERATOR KEANGGOTAAN")
print("======================================")

# Input beberapa data yang dipisahkan dengan koma
input_data_1008 = input("Masukkan beberapa angka, pisahkan dengan koma: ")

# Mengubah input menjadi list integer
data_1008 = [int(angka.strip()) for angka in input_data_1008.split(',')]

nilai_dicari_1008 = int(input("Masukkan angka yang ingin dicari: "))

# Operator in
hasil_1008 = nilai_dicari_1008 in data_1008
print("\nOperator keanggotaan IN")
print(nilai_dicari_1008,"in",data_1008,"=",hasil_1008)

# Operator not in
hasil_1008 = nilai_dicari_1008 not in data_1008
print("\nOperator keanggotaan NOT IN")
print(nilai_dicari_1008,"not in",data_1008,"=",hasil_1008)

print("======================================")
print("2. OPERATOR IDENTIAS")
print("======================================")

# objek1 menggunakan list dari input perngguna
object1_1008 = data_1008

# objek2 merujuk pada objek yang sama dengan objek1
object2_1008 = object1_1008

# objek3 memiliki isi sama, tetapi merupakan objek baru
object3_1008 = data_1008.copy()

print("object1 =",object1_1008)
print("object2 =",object2_1008)
print("object3 =",object3_1008)

# Operator is
hasil_1008 = object1_1008 is object2_1008
print("\nOperator identitas IS")
print("objek1 is objek2 =",hasil_1008)

# Operator is not
hasil_1008 = object1_1008 is not object2_1008
print("\nOperator identitas IS NOT")
print("objek1 is not objek2 =",hasil_1008)

# Membandingkan identitas dan nilai
print("\nPerbandingan identitas dan nilai")
print("objek1 is objek3 =",object1_1008 is object3_1008)
print("objek1 == objek3 =",object1_1008 == object3_1008)
```

**Penjelasan Program**

**Bagian 1: Operator Keanggotaan**
1. Program meminta pengguna memasukkan beberapa angka dalam satu baris dipisahkan koma, disimpan pada `input_data_1008`.
2. Input tersebut dipecah menggunakan `split(',')`, dibersihkan spasinya dengan `strip()`, lalu diubah menjadi integer dengan `int()` melalui list comprehension, sehingga terbentuk list angka pada variabel `data_1008`.
3. Pengguna memasukkan satu angka lagi (`nilai_dicari_1008`) yang ingin dicari keberadaannya di dalam list.
4. Operator in memeriksa apakah `nilai_dicari_1008` terdapat di dalam `data_1008`, hasilnya `True` atau `False`.
5. Operator not in memeriksa kebalikannya, yaitu apakah angka tersebut tidak terdapat di dalam list.

**Bagian 2: Operator Identitas**
1. `object1_1008` diisi dengan `data_1008`.
2. `object2_1008` diisi dengan `object1_1008`, sehingga keduanya menunjuk ke objek yang sama persis di memori.
3. `object3_1008` dibuat dengan `data_1008.copy()`, isinya sama tetapi merupakan objek baru di memori yang berbeda.
4. Operator is memeriksa apakah dua variabel merujuk pada objek yang sama di memori; `object1_1008 is object2_1008` bernilai `True` karena keduanya memang objek yang sama.
5. Operator is not memeriksa kebalikannya, hasilnya `False` untuk pasangan objek yang sama.
6. Perbandingan `object1_1008 is object3_1008` bernilai `False` karena keduanya objek berbeda di memori, sedangkan `object1_1008 == object3_1008` bernilai `True` karena `==` hanya membandingkan isi/nilai, bukan lokasi memori.


### 2.2.5 Program Logika "B_2611531008"

```python
# Buat file dengan nama logika_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: a1_1008
# Program ini menggunakan fungsi input()
# Program operator logika dalam Python

# Masukkan nilai boolean
# Input tidak peka terhadap huruf besar dan kecil
a1_1008 = input("Input nilai boolean-1 (true/false): ").strip().lower() == "true"
a2_1008 = input("Input nilai boolean-2 (ture/false): ").strip().lower() == "true"

print("\nA1 =", a1_1008)
print("\nA2 =", a2_1008)

# Konjugsi: bernilai True jika keduanya True
hasil = a1_1008 and a2_1008
print("\nKonjugsi (AND)")
print("A1 and A2 =", hasil)

# Disjungsi: bernilai True jika salah satunya True
hasil = a1_1008 or a2_1008
print("\nDisjungsi (OR)")
print("not A1 or A2 =", hasil)

# Negasi A1: membalik nilai A1
hasil = not a1_1008
print("\nNegasi A1 (NOT)")
print("not A1 =", hasil)

# Negasi A2: membalik nilai A2
hasil = not a2_1008
print("\nNegasi A2 (NOT)")
print("not A2 =", hasil)

# XOR: bernilai True jika kedua nilai berbeda
hasil = a1_1008 != a2_1008
print("\nDisjungsi Eksklusif (XOR)")
print("A1 XOR A2 =", hasil)
```

**Penjelasan Program**

1. Program meminta pengguna memasukkan dua nilai boolean dalam bentuk teks "true"/"false" melalui `input()`.
2. Input dibersihkan dengan `.strip()` dan diubah ke huruf kecil dengan `.lower()`, lalu dibandingkan dengan `== "true"`, sehingga hasilnya langsung berupa nilai boolean `True` atau `False` tanpa perlu `if` tambahan.
3. Kedua nilai boolean (`a1_1008` dan `a2_1008`) ditampilkan terlebih dahulu sebelum diproses.
4. Operator konjungsi (and) menghasilkan `True` hanya jika `a1_1008` dan `a2_1008` sama-sama `True`.
5. Operator disjungsi (or) menghasilkan `True` apabila salah satu dari `a1_1008` atau `a2_1008` bernilai `True`.
6. Operator negasi (not) diterapkan terpisah pada `a1_1008` dan `a2_1008`, berfungsi membalik nilai boolean dari `True` menjadi `False` atau sebaliknya.
7. Operator disjungsi eksklusif (XOR) menggunakan `!=` sebagai alternatif, karena Python tidak memiliki operator XOR khusus untuk boolean; hasilnya `True` jika `a1_1008` dan `a2_1008` berbeda nilai, dan `False` jika sama.


### 2.2.6 Program Perbandingan "B_2611531008"

```python
# ============================================================
# Tugas 3 - Studi Kasus Terpadu
# Sistem Simulasi Transaksi dan Validasi Akses Toko
# Nama Berkas : tugas3_1008.py
# NIM         : ****1008
# ============================================================

print("=== SISTEM TRANSAKSI TOKO ===")

# ---------- INPUT DATA ----------
nama_1008 = input("Masukkan Nama Pelanggan : ")
status_1008 = input("Masukkan Status Pelanggan (member/nonmember) : ")
total_1008 = int(input("Masukkan Total Belanja : "))
jumlah_1008 = int(input("Masukkan Jumlah Barang : "))
promo_1008 = input("Masukkan Kode Promo : ")

print("\n=== DATA TRANSAKSI ===")
print("Nama Pelanggan       :", nama_1008)
print("Status Pelanggan     :", status_1008)
print(f"Total Belanja        : Rp{total_1008}")
print("Jumlah Barang        :", jumlah_1008)
print("Kode Promo           :", promo_1008)

# ---------- OPERATOR KEANGGOTAAN (in) ----------
kode_promo_1008 = ["HEMAT10", "HEMAT20", "HIDUPMAHASISWA", "HIDUPFTI"]

# ---------- OPERATOR PERBANDINGAN ----------
syarat_total_1008 = total_1008 >= 200000        # >=
syarat_jumlah_1008 = jumlah_1008 >= 3           # >=
status_valid_1008 = status_1008 == "member"     # ==
promo_valid_1008 = promo_1008 in kode_promo_1008  # in (keanggotaan)

print("\n=== HASIL VALIDASI ===")
print("Belanja >= Rp200000        :", syarat_total_1008)
print("Jumlah Barang >= 3         :", syarat_jumlah_1008)
print("Status Member              :", status_valid_1008)
print("Kode Promo Tersedia        :", promo_valid_1008)

# ---------- OPERATOR LOGIKA ----------
dapat_diskon_1008 = syarat_jumlah_1008 or syarat_total_1008   # OR
dapat_promo_1008 = promo_valid_1008 and status_valid_1008     # AND

print("Mendapatkan Diskon         :", dapat_diskon_1008)
print("Mendapatkan Promo          :", dapat_promo_1008)

# ---------- OPERATOR ARITMATIKA & ASSIGNMENT ----------
diskon_persen_1008 = 0.05

nilai_diskon_1008 = 0
if dapat_diskon_1008:
    nilai_diskon_1008 = total_1008 * diskon_persen_1008   # perkalian (*)

total_bayar_1008 = total_1008
total_bayar_1008 -= nilai_diskon_1008                     # assignment pengurangan (-=)

rata_harga_1008 = total_bayar_1008 / jumlah_1008           # pembagian (/)

print("\n=== HASIL PERHITUNGAN ===")
print(f"Diskon                      : Rp{nilai_diskon_1008}")
print(f"Total Pembayaran            : Rp{total_bayar_1008}")
print(f"Rata-rata Harga Barang      : Rp{rata_harga_1008}")

# ---------- OPERATOR BITWISE (HAK AKSES) ----------
# Bit 0 = Member Access
# Bit 1 = Free Shipping Access (syarat total belanja terpenuhi)
# Bit 2 = Diskon Access (syarat jumlah barang terpenuhi)
# Bit 3 = Promo Access

kode_transaksi_1008 = (
    int(status_valid_1008) << 0
    | int(syarat_total_1008) << 1
    | int(syarat_jumlah_1008) << 2
    | int(promo_valid_1008) << 3
)

free_shipping_1008 = syarat_total_1008 and syarat_jumlah_1008  # AND

print("\n=== HAK AKSES PELANGGAN ===")
print(f"Kode Hak Akses              : {format(kode_transaksi_1008, '04b')}")
print("Member Access               :", status_valid_1008)
print("Promo Access                :", promo_valid_1008)
print("Free Shipping Access        :", free_shipping_1008)

# Kode referensi pembanding (tanpa syarat jumlah barang)
kode_referensi_1008 = (
    int(status_valid_1008) << 0
    | int(syarat_total_1008) << 1
    | int(promo_valid_1008) << 3
)

print("\n=== OPERASI BITWISE ===")
print("=== Kode Status Transaksi ===")
print(f"{format(int(status_valid_1008) << 0, '04b')} | "
      f"{format(int(syarat_total_1008) << 1, '04b')} | "
      f"{format(int(syarat_jumlah_1008) << 2, '04b')} | "
      f"{format(int(promo_valid_1008) << 3, '04b')}")
print(f"Kode Biner   : {format(kode_transaksi_1008, '04b')}")
print("Kode Desimal :", kode_transaksi_1008)

# Cek Member (AND dengan mask bit 0)
print("\n=== Pemeriksaan Status ===")
print("Cek Member")
mask_member_1008 = int(status_valid_1008) << 0
hasil_member_1008 = kode_transaksi_1008 & mask_member_1008
print(f"{format(kode_transaksi_1008, '04b')} & {format(mask_member_1008, '04b')}")
print(f"Hasil Biner   : {format(hasil_member_1008, '04b')}")
print(f"Hasil Desimal : {hasil_member_1008}")

# Cek Promo (AND dengan mask bit 3)
print("\nCek Promo")
mask_promo_1008 = int(promo_valid_1008) << 3
hasil_promo_1008 = kode_transaksi_1008 & mask_promo_1008
print(f"{format(kode_transaksi_1008, '04b')} & {format(mask_promo_1008, '04b')}")
print(f"Hasil Biner   : {format(hasil_promo_1008, '04b')}")
print(f"Hasil Desimal : {hasil_promo_1008}")

# Perbandingan status transaksi vs referensi (XOR)
print("\n=== Perbandingan Status ===")
hasil_xor_1008 = kode_transaksi_1008 ^ kode_referensi_1008
print(f"Kode Transaksi : {format(kode_transaksi_1008, '04b')}")
print(f"Kode Referensi : {format(kode_referensi_1008, '04b')}")
print(f"{format(kode_transaksi_1008, '04b')} ^ {format(kode_referensi_1008, '04b')}")
print(f"Hasil Biner   : {format(hasil_xor_1008, '04b')}")
print(f"Hasil Desimal : {hasil_xor_1008}")

# Operasi shift (contoh: naik level membership)
print("\n=== Shift ===")
hasil_shift_1008 = kode_transaksi_1008 << 1
print(f"{format(kode_transaksi_1008, '04b')} << 1")
print(f"Hasil Biner   : {format(hasil_shift_1008, '04b')}")
print(f"Hasil Desimal : {hasil_shift_1008}")

print("=== SELESAI ===")
```

### Penjelasan Program (Bentuk Pointer)

#### 1. Input Data

* Program meminta pengguna memasukkan:

  * Nama pelanggan.
  * Status pelanggan (`member` atau `nonmember`).
  * Total belanja.
  * Jumlah barang yang dibeli.
  * Kode promo.

#### 2. Menampilkan Data Transaksi

* Semua data yang telah diinput ditampilkan kembali sebagai informasi transaksi.

#### 3. Validasi Data

* `syarat_total_1008 = total_1008 >= 200000`

  * Memeriksa apakah total belanja minimal Rp200.000.
* `syarat_jumlah_1008 = jumlah_1008 >= 3`

  * Memeriksa apakah jumlah barang minimal 3.
* `status_valid_1008 = status_1008 == "member"`

  * Memeriksa apakah pelanggan berstatus member.
* `promo_valid_1008 = promo_1008 in kode_promo_1008`

  * Memeriksa apakah kode promo terdapat dalam daftar promo yang tersedia.

#### 4. Penentuan Diskon dan Promo

* Pelanggan mendapatkan diskon jika:

  * Jumlah barang ≥ 3 **atau**
  * Total belanja ≥ Rp200.000.
* Pelanggan mendapatkan promo jika kode promo yang dimasukkan valid.

#### 5. Perhitungan Pembayaran

* Diskon yang digunakan adalah 5% (`0,05`).
* Besar diskon dihitung dengan:

  * `diskon = total_belanja × 5%`
* Total pembayaran:

  * `total_belanja − diskon`
* Rata-rata harga per barang:

  * `total_pembayaran ÷ jumlah_barang`

#### 6. Hak Akses Pelanggan

* `Member Access`

  * Bernilai `True` jika pelanggan adalah member.
* `Promo Access`

  * Bernilai `True` jika kode promo valid.
* `Free Shipping Access`

  * Belum ditentukan pada program (masih berupa `...`).

#### 7. Operasi Bitwise

* Program mengubah status transaksi menjadi kode biner.
* Setiap kondisi memiliki posisi bit:

  * Bit 0 = Status Member.
  * Bit 1 = Total Belanja ≥ Rp200.000.
  * Bit 2 = Jumlah Barang ≥ 3.
  * Bit 3 = Kode Promo Valid.
* Jika semua kondisi bernilai `True`, maka:

  * Biner = `1111`
  * Desimal = `15`

#### 8. Pemeriksaan Status dengan Operator AND (`&`)

* Digunakan untuk mengecek apakah suatu status aktif.
* Cek Member:

  * `1111 & 0001 = 0001`
  * Hasil menunjukkan pelanggan adalah member.
* Cek Promo:

  * `1111 & 1000 = 1000`
  * Hasil menunjukkan promo valid.

#### 9. Perbandingan Status dengan Operator XOR (`^`)

* Digunakan untuk membandingkan dua kode status.
* Bit yang berbeda akan menghasilkan nilai `1`.
* Contoh:

  * `1111 ^ 1011 = 0100`
* Hasil desimal = `4`.

#### 10. Operasi Shift Kiri (`<<`)

* Digunakan untuk menggeser seluruh bit satu langkah ke kiri.
* Contoh:

  * `1111 << 1 = 11110`
* Nilai desimal berubah dari `15` menjadi `30`.
* Pergeseran kiri 1 bit sama dengan mengalikan nilai dengan 2.
