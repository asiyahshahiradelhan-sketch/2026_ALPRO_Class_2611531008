## 2.1 Dasar Teori
# Landasan Teori

Algoritma pemrograman merupakan urutan langkah logis dan sistematis untuk menyelesaikan suatu masalah, yang kemudian diterjemahkan ke dalam bahasa pemrograman agar dapat dieksekusi oleh komputer. Salah satu bahasa pemrograman yang umum digunakan untuk pengenalan algoritma adalah **Python**, yaitu bahasa tingkat tinggi yang bersifat *interpreted*, memiliki sintaks sederhana dan mudah dibaca, serta mendukung berbagai paradigma pemrograman seperti prosedural dan berorientasi objek. Elemen dasar dalam Python meliputi variabel, tipe data, operator, struktur kontrol (percabangan dan perulangan), fungsi, serta input/output, yang menjadi fondasi dalam mengimplementasikan suatu algoritma.

Untuk menulis dan menjalankan kode Python, digunakan **Visual Studio Code (VS Code)**, yaitu *source code editor* ringan yang dikembangkan oleh Microsoft. VS Code mendukung berbagai ekstensi, seperti ekstensi Python, yang menghadirkan fitur *syntax highlighting*, *auto-completion*, debugging, serta terminal terintegrasi sehingga proses menulis dan menguji kode dapat dilakukan dalam satu aplikasi. Selain itu, VS Code juga terintegrasi dengan sistem kontrol versi Git, sehingga pengelolaan kode dapat dilakukan langsung dari editor tanpa perlu berpindah aplikasi.

Sistem kontrol versi yang digunakan adalah **Git**, yaitu sistem yang melacak setiap perubahan pada kode sumber melalui mekanisme *commit*, sedangkan **GitHub** merupakan platform daring untuk menyimpan *repository* Git tersebut sehingga memungkinkan kolaborasi, pencadangan, dan pengelolaan riwayat proyek secara terpusat. Konsep penting dalam Git dan GitHub antara lain *repository* sebagai folder proyek yang dilacak, *commit* sebagai catatan perubahan, *branch* sebagai cabang pengembangan, serta *push* dan *pull* sebagai proses mengirim dan mengambil data antara repository lokal dan remote. Secara keseluruhan, ketiga komponen ini saling berkaitan dalam praktikum: Python berperan sebagai bahasa untuk menulis algoritma, VS Code sebagai alat untuk menulis dan menjalankan kode, sedangkan Git dan GitHub berperan dalam mengelola versi serta mendukung kolaborasi terhadap kode yang telah dibuat.

1. Python<br>
    Python merupakan bahasa pemrograman tingkat tinggi yang memiliki sintaks sederhana sehingga mudah dipahami oleh pemula. Python dapat digunakan untuk berbagai kebutuhan, seperti pembuatan aplikasi, pengolahan data, dan pengembangan perangkat lunak.<br>
2. Visual Studio Code<br>
    Visual Studio Code (VS Code) merupakan code editor yang digunakan untuk menulis, mengedit, dan menjalankan kode program. VS Code mendukung berbagai bahasa pemrograman, termasuk Python, serta menyediakan fitur seperti terminal, debugging, dan integrasi dengan Git.<br>
3. Git<br>
    Git merupakan sistem version control yang digunakan untuk mencatat perubahan pada kode program. Dengan Git, programmer dapat mengelola berbagai versi program dan mengetahui perubahan yang telah dilakukan.
4. GitHub<br>
    GitHub merupakan platform berbasis daring yang digunakan untuk menyimpan dan mengelola repository Git. GitHub dapat membantu programmer dalam melakukan pencadangan kode, mengakses proyek dari berbagai perangkat, serta bekerja sama dalam pengembangan program.
5. Repository<br>
    Repository merupakan tempat penyimpanan proyek yang berisi kode program dan berbagai berkas pendukung. Repository dapat disimpan secara lokal pada komputer maupun secara daring pada GitHub.
6. Commit dan Push<br>
    Commit digunakan untuk menyimpan perubahan yang telah dilakukan pada repository lokal. Push digunakan untuk mengirim perubahan dari repository lokal ke repository yang berada di GitHub.
